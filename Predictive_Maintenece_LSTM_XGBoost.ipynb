```python
# ============================================================
#   HYBRID XGBoost–LSTM  |  RUL Prediction  |  ALL FD Datasets
#   FINAL FIXED VERSION — Google Colab T4 GPU
#
#   ✅ UPLOAD YOUR OWN ZIP:
#      When the cell runs, a file-picker button will appear.
#      Click it and select your CMAPSSData.zip from your computer.
#      The script handles extraction automatically.
#
#   💡 How it works:
#      Step 1 → Check if files already exist on disk (skip if yes)
#      Step 2 → Try auto-download from NASA URL
#      Step 3 → If download fails → show upload button (Colab)
#                                  → print manual instructions (local)
# ============================================================

# ── 0. Install ───────────────────────────────────────────────
import subprocess
subprocess.run(['pip', 'install', 'xgboost', 'scikit-learn',
                'pandas', 'numpy', 'scipy', 'joblib', '-q'])

# ── 1. Imports ───────────────────────────────────────────────
import os, warnings, joblib, zipfile, urllib.request, shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader, random_split
from xgboost import XGBRegressor
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import mean_absolute_error, r2_score
from scipy.optimize import minimize_scalar
warnings.filterwarnings('ignore')
torch.manual_seed(42)
np.random.seed(42)

# ── 2. Per-Dataset Config ────────────────────────────────────
DATASET_CFG = {
    'FD001': {'n_clusters': 1, 'epochs': 100, 'xgb_trees': 1500},
    'FD002': {'n_clusters': 6, 'epochs': 120, 'xgb_trees': 2000},
    'FD003': {'n_clusters': 1, 'epochs': 100, 'xgb_trees': 1500},
    'FD004': {'n_clusters': 6, 'epochs': 120, 'xgb_trees': 2000},
}

# ── 3. Global Config ─────────────────────────────────────────
SEQUENCE_LENGTH = 50
RUL_CAP         = 125
BATCH_SIZE      = 256
LR              = 1e-3
HIDDEN_SIZE     = 128
NUM_LAYERS      = 2
DROPOUT         = 0.2
DEVICE          = 'cuda' if torch.cuda.is_available() else 'cpu'

print(f"✅ Device : {DEVICE}")
if DEVICE == 'cuda':
    print(f"   GPU   : {torch.cuda.get_device_name(0)}")
    print(f"   VRAM  : {torch.cuda.get_device_properties(0).total_memory/1e9:.1f} GB")

# ── 4. Dataset Download / Upload ─────────────────────────────
os.makedirs('CMAPSSData', exist_ok=True)

ZIP_FILENAME    = "CMAPSSData.zip"
EXTRACT_DIR     = "CMAPSSData"
NASA_URL        = (
    "https://data.nasa.gov/api/views/ff5v-kuh6/files/"
    "a9b11d15-5b5e-4f5e-8e4e-db67c4ea2a46"
    "?download=true&filename=CMAPSSData.zip"
)

def find_file(name, search_root='.'):
    """Recursively find a file by name. Returns full path or None."""
    for root, _, files in os.walk(search_root):
        if name in files:
            return os.path.join(root, name)
    return None


def extract_zip(zip_path, dest_dir):
    """Extract a ZIP file and print each extracted member."""
    print(f"📦 Extracting  {zip_path}  →  {dest_dir}/")
    with zipfile.ZipFile(zip_path, 'r') as zf:
        members = zf.namelist()
        for i, member in enumerate(members, 1):
            zf.extract(member, dest_dir)
            print(f"\r   {i}/{len(members)}: {member:<50}", end="", flush=True)
    print(f"\n✅ Extracted {len(members)} files.")


def _is_colab():
    """Detect if running inside Google Colab."""
    try:
        import google.colab  # noqa: F401
        return True
    except ImportError:
        return False


def upload_and_extract():
    """
    ─────────────────────────────────────────────────────────
    MANUAL UPLOAD PATH
    ─────────────────────────────────────────────────────────
    Called when auto-download fails.

    In Google Colab:
        A file picker widget appears.
        Select your  CMAPSSData.zip  and click Upload.
        The script extracts it automatically.

    Locally:
        Place CMAPSSData.zip in the same folder as this script,
        then re-run. Instructions are printed to the console.
    ─────────────────────────────────────────────────────────
    """
    if _is_colab():
        print("\n" + "="*60)
        print("  📂  UPLOAD YOUR ZIP FILE")
        print("  ─────────────────────────────────────────────────────")
        print("  1. Click the  'Choose Files'  button that appears below.")
        print("  2. Navigate to your  CMAPSSData.zip  file.")
        print("  3. Click Open / Upload.")
        print("  4. Wait for the upload progress bar to complete.")
        print("="*60 + "\n")

        from google.colab import files
        uploaded = files.upload()          # ← shows the file-picker widget

        if not uploaded:
            raise RuntimeError("No file was uploaded. Please re-run and upload CMAPSSData.zip.")

        for uploaded_name, content in uploaded.items():
            dest_path = os.path.join('.', uploaded_name)
            with open(dest_path, 'wb') as f:
                f.write(content)
            print(f"✅ Received: {uploaded_name}  ({len(content)/1e6:.1f} MB)")

            # If user uploaded the zip directly
            if uploaded_name.lower().endswith('.zip'):
                extract_zip(dest_path, EXTRACT_DIR)

            # If user uploaded individual .txt files
            elif uploaded_name.endswith('.txt'):
                dest = os.path.join(EXTRACT_DIR, uploaded_name)
                shutil.move(dest_path, dest)
                print(f"   Moved → {dest}")

    else:
        # ── Running locally (not Colab) ──────────────────────────────────────
        print("\n" + "="*60)
        print("  ❌  Auto-download failed. Manual steps required:")
        print("  ─────────────────────────────────────────────────────")
        print("  Option A — Download the ZIP manually:")
        print("    1. Go to: https://data.nasa.gov/dataset/C-MAPSS")
        print("    2. Download  CMAPSSData.zip")
        print(f"   3. Place it here:  {os.path.abspath('.')}/CMAPSSData.zip")
        print("    4. Re-run this script.")
        print()
        print("  Option B — Place extracted files directly:")
        print(f"   Copy these files into:  {os.path.abspath(EXTRACT_DIR)}/")
        print("      train_FD001.txt   test_FD001.txt   RUL_FD001.txt")
        print("      train_FD002.txt   test_FD002.txt   RUL_FD002.txt")
        print("      train_FD003.txt   test_FD003.txt   RUL_FD003.txt")
        print("      train_FD004.txt   test_FD004.txt   RUL_FD004.txt")
        print("="*60 + "\n")
        raise SystemExit(1)


def download_cmapss():
    """
    Full dataset acquisition pipeline.

    Priority 1 → Files already on disk      → skip everything
    Priority 2 → Auto-download from NASA URL → urllib
    Priority 3 → Manual upload / placement   → upload_and_extract()
    """

    # ── Priority 1: already extracted? ───────────────────────────────────────
    if find_file('train_FD001.txt'):
        print("✅ Dataset already present on disk — skipping download.\n")
        return

    # ── Priority 2: ZIP already on disk but not extracted? ───────────────────
    if os.path.isfile(ZIP_FILENAME) and os.path.getsize(ZIP_FILENAME) > 50_000:
        print(f"✅ Found {ZIP_FILENAME} on disk — extracting …")
        try:
            extract_zip(ZIP_FILENAME, EXTRACT_DIR)
            if find_file('train_FD001.txt'):
                return
        except zipfile.BadZipFile:
            print(f"⚠️  {ZIP_FILENAME} appears corrupted. Will try re-download.")
            os.remove(ZIP_FILENAME)

    # ── Priority 3: Try NASA URL download ────────────────────────────────────
    print("⬇️  Attempting auto-download from data.nasa.gov …")
    print(f"   URL: {NASA_URL[:70]}…\n")

    download_ok = False
    try:
        def _progress(block_num, block_size, total_size):
            done = block_num * block_size
            if total_size > 0:
                pct  = min(done / total_size * 100, 100)
                bar  = "█" * int(pct / 2) + "░" * (50 - int(pct / 2))
                print(f"\r   [{bar}] {pct:5.1f}%  {done//1024:,}/{total_size//1024:,} KB",
                      end="", flush=True)

        urllib.request.urlretrieve(NASA_URL, ZIP_FILENAME, _progress)
        print(f"\n✅ Download complete: {ZIP_FILENAME}")
        download_ok = True

    except Exception as exc:
        print(f"\n⚠️  Auto-download failed: {exc.__class__.__name__}: {exc}")
        download_ok = False

    if download_ok:
        try:
            extract_zip(ZIP_FILENAME, EXTRACT_DIR)
            if find_file('train_FD001.txt'):
                return
            else:
                print("⚠️  Extraction succeeded but train_FD001.txt not found inside ZIP.")
        except zipfile.BadZipFile:
            print("⚠️  Downloaded file is not a valid ZIP. Falling back to manual upload.")
            os.remove(ZIP_FILENAME)

    # ── Priority 4: Manual upload / instructions ──────────────────────────────
    upload_and_extract()


# ── Run dataset acquisition ───────────────────────────────────
download_cmapss()

# ── Verify all required files exist before proceeding ─────────
required = [f'{split}_{fd}.txt'
            for fd in ['FD001','FD002','FD003','FD004']
            for split in ['train','test','RUL']]
missing = [f for f in required if not find_file(f)]
if missing:
    print(f"\n⚠️  Still missing {len(missing)} files after setup:")
    for m in missing:
        print(f"   • {m}")
    print("\nOnly datasets with all 3 files present will be processed.\n")
else:
    print(f"\n✅ All {len(required)} dataset files verified.\n")


# ── 5. Preprocessing ─────────────────────────────────────────
COLUMNS      = ['engine_id','cycle','op1','op2','op3'] + [f's{i}' for i in range(1,22)]
DROP_SENSORS = ['s1','s5','s6','s10','s16','s18','s19']

def load_data(fd):
    tp = find_file(f'train_{fd}.txt')
    ep = find_file(f'test_{fd}.txt')
    rp = find_file(f'RUL_{fd}.txt')
    if not all([tp, ep, rp]):
        raise FileNotFoundError(f"Missing files for {fd}: train={tp}, test={ep}, RUL={rp}")
    train_df = pd.read_csv(tp, sep=r'\s+', header=None, names=COLUMNS).dropna(axis=1)
    test_df  = pd.read_csv(ep, sep=r'\s+', header=None, names=COLUMNS).dropna(axis=1)
    rul_df   = pd.read_csv(rp, sep=r'\s+', header=None, names=['RUL']).dropna(axis=1)
    for df in [train_df, test_df]:
        df.drop(columns=[c for c in DROP_SENSORS if c in df.columns], inplace=True)
    max_cyc         = train_df.groupby('engine_id')['cycle'].max().reset_index()
    max_cyc.columns = ['engine_id','max_cycle']
    train_df        = train_df.merge(max_cyc, on='engine_id')
    train_df['RUL'] = (train_df['max_cycle'] - train_df['cycle']).clip(upper=RUL_CAP)
    train_df.drop(columns=['max_cycle'], inplace=True)
    return train_df, test_df, rul_df

def cluster_operating_conditions(train_df, test_df, n_clusters):
    op_cols = ['op1','op2','op3']
    if n_clusters > 1:
        km = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
        train_df['op_cluster'] = km.fit_predict(train_df[op_cols])
        test_df['op_cluster']  = km.predict(test_df[op_cols])
    else:
        train_df['op_cluster'] = 0
        test_df['op_cluster']  = 0
    return train_df, test_df

def engineer_features(df):
    sensor_cols = [c for c in df.columns
                   if c.startswith('s') and '_' not in c and c != 'op_cluster']
    for col in sensor_cols:
        g = df.groupby('engine_id')[col]
        df[f'{col}_rmean'] = g.transform(lambda x: x.rolling(10, min_periods=1).mean())
        df[f'{col}_rstd']  = g.transform(lambda x: x.rolling(10, min_periods=1).std().fillna(0))
        df[f'{col}_slope'] = g.transform(lambda x: x.diff().fillna(0))
        df[f'{col}_ewm']   = g.transform(lambda x: x.ewm(span=10, adjust=False).mean())
    return df

def normalize_by_cluster(train_df, test_df):
    exclude   = {'engine_id','cycle','RUL','op_cluster'}
    feat_cols = [c for c in train_df.columns if c not in exclude]
    scaler_map = {}
    for cl in train_df['op_cluster'].unique():
        sc   = MinMaxScaler()
        idx  = train_df['op_cluster'] == cl
        train_df.loc[idx, feat_cols] = sc.fit_transform(train_df.loc[idx, feat_cols])
        tidx = test_df['op_cluster'] == cl
        if tidx.any():
            tcols = [c for c in feat_cols if c in test_df.columns]
            test_df.loc[tidx, tcols] = sc.transform(test_df.loc[tidx, tcols])
        scaler_map[cl] = sc
    return train_df, test_df, scaler_map, feat_cols

def make_sequences(df, feat_cols):
    X, y = [], []
    for eid in df['engine_id'].unique():
        edf   = df[df['engine_id']==eid].reset_index(drop=True)
        feats = edf[feat_cols].values.astype(np.float32)
        labels= edf['RUL'].values.astype(np.float32)
        for i in range(SEQUENCE_LENGTH, len(edf)):
            X.append(feats[i-SEQUENCE_LENGTH:i])
            y.append(labels[i])
    return np.array(X, dtype=np.float32), np.array(y, dtype=np.float32)

def make_test_sequences(df, feat_cols):
    X = []
    for eid in df['engine_id'].unique():
        edf   = df[df['engine_id']==eid].reset_index(drop=True)
        feats = edf[feat_cols].values.astype(np.float32)
        if len(feats) >= SEQUENCE_LENGTH:
            X.append(feats[-SEQUENCE_LENGTH:])
        else:
            pad = np.zeros((SEQUENCE_LENGTH-len(feats), feats.shape[1]), dtype=np.float32)
            X.append(np.vstack([pad, feats]))
    return np.array(X, dtype=np.float32)

# ── 6. LSTM with Attention ───────────────────────────────────
class Attention(nn.Module):
    def __init__(self, hidden_size):
        super().__init__()
        self.attn = nn.Linear(hidden_size, 1)
    def forward(self, lstm_out):
        w = torch.softmax(self.attn(lstm_out), dim=1)
        return (w * lstm_out).sum(dim=1)

class LSTMModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.lstm      = nn.LSTM(input_size, HIDDEN_SIZE, NUM_LAYERS,
                                 batch_first=True, dropout=DROPOUT)
        self.attention = Attention(HIDDEN_SIZE)
        self.bn        = nn.BatchNorm1d(HIDDEN_SIZE)
        self.dropout   = nn.Dropout(0.3)
        self.fc1       = nn.Linear(HIDDEN_SIZE, 64)
        self.fc2       = nn.Linear(64, 1)
        self.relu      = nn.ReLU()
    def forward(self, x):
        out, _ = self.lstm(x)
        ctx    = self.attention(out)
        ctx    = self.dropout(self.bn(ctx))
        return self.fc2(self.relu(self.fc1(ctx))).squeeze(-1)

def train_lstm(model, train_loader, val_loader, epochs):
    model.to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=1e-4)
    criterion = nn.HuberLoss(delta=10.0)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
                    optimizer, T_max=epochs, eta_min=1e-5)
    best_val, best_state = float('inf'), None
    history = {'train': [], 'val': []}
    for epoch in range(1, epochs+1):
        model.train()
        tl = 0.0
        for xb, yb in train_loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            tl += loss.item() * len(xb)
        tl /= len(train_loader.dataset)
        model.eval()
        vl = 0.0
        with torch.no_grad():
            for xb, yb in val_loader:
                xb, yb = xb.to(DEVICE), yb.to(DEVICE)
                vl += criterion(model(xb), yb).item() * len(xb)
        vl /= len(val_loader.dataset)
        scheduler.step()
        history['train'].append(tl)
        history['val'].append(vl)
        if vl < best_val:
            best_val   = vl
            best_state = {k: v.clone() for k, v in model.state_dict().items()}
        if epoch % 10 == 0:
            print(f"  Epoch {epoch:3d}/{epochs} | "
                  f"Train: {tl:.4f} | Val: {vl:.4f} | "
                  f"LR: {optimizer.param_groups[0]['lr']:.6f}")
    model.load_state_dict(best_state)
    print(f"  ✅ Best Val Loss: {best_val:.4f}")
    return model, history

def predict_lstm(model, loader):
    model.eval()
    preds = []
    with torch.no_grad():
        for xb, *_ in loader:
            preds.append(model(xb.to(DEVICE)).cpu().numpy())
    return np.concatenate(preds)

# ── 7. XGBoost Features ──────────────────────────────────────
def get_tabular_features(X):
    return np.hstack([
        X[:, -1, :],
        X.mean(axis=1),
        X.std(axis=1),
        X[:, -1, :] - X[:, 0, :],
        np.percentile(X, 25, axis=1),
        np.percentile(X, 75, axis=1),
        np.percentile(X, 75, axis=1) - np.percentile(X, 25, axis=1),
        X.max(axis=1) - X.min(axis=1),
    ])

# ── 8. Evaluate ──────────────────────────────────────────────
def evaluate(y_true, y_pred, label):
    r  = np.sqrt(np.mean((y_true - y_pred)**2))
    m  = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"  {'─'*40}")
    print(f"  {label:<25} RMSE:{r:7.4f}  MAE:{m:7.4f}  R²:{r2:.4f}")
    return {'RMSE': r, 'MAE': m, 'R2': r2}

# ── 9. Plot ───────────────────────────────────────────────────
def plot_results(fd, history, results, y_test,
                 pred_lstm, pred_xgb, pred_hybrid, alpha):
    fig, axes = plt.subplots(2, 3, figsize=(20, 12))
    fig.suptitle(f'Hybrid XGBoost–LSTM  |  {fd}  |  RUL Prediction',
                 fontsize=16, fontweight='bold')
    ax = axes[0,0]
    ax.plot(history['train'], label='Train', color='royalblue', lw=2)
    ax.plot(history['val'],   label='Val',   color='tomato',    lw=2)
    ax.set_title('LSTM Loss (Huber)'); ax.set_xlabel('Epoch')
    ax.legend(); ax.grid(alpha=0.3)
    ax = axes[0,1]
    names  = list(results.keys())
    rmses  = [results[k]['RMSE'] for k in names]
    colors = ['royalblue','seagreen','darkorange']
    bars   = ax.bar(names, rmses, color=colors, edgecolor='white', width=0.5)
    for bar, v in zip(bars, rmses):
        ax.text(bar.get_x()+bar.get_width()/2,
                bar.get_height()+0.2, f'{v:.2f}',
                ha='center', fontweight='bold', fontsize=12)
    ax.set_title('RMSE Comparison'); ax.set_ylabel('RMSE'); ax.grid(axis='y', alpha=0.3)
    ax = axes[0,2]
    r2s  = [results[k]['R2'] for k in names]
    bars = ax.bar(names, r2s, color=colors, edgecolor='white', width=0.5)
    for bar, v in zip(bars, r2s):
        ax.text(bar.get_x()+bar.get_width()/2,
                bar.get_height()+0.005, f'{v:.3f}',
                ha='center', fontweight='bold', fontsize=12)
    ax.set_title('R² Comparison'); ax.set_ylabel('R²')
    ax.set_ylim(0, 1.1); ax.grid(axis='y', alpha=0.3)
    mn, mx = float(y_test.min()), float(y_test.max())
    ax = axes[1,0]
    ax.scatter(y_test, pred_lstm, alpha=0.5, color='royalblue', s=20)
    ax.plot([mn,mx],[mn,mx],'r--',lw=2)
    ax.set_title(f'LSTM  RMSE={results["LSTM"]["RMSE"]:.2f}')
    ax.set_xlabel('Actual RUL'); ax.set_ylabel('Predicted RUL'); ax.grid(alpha=0.3)
    ax = axes[1,1]
    ax.scatter(y_test, pred_hybrid, alpha=0.5, color='darkorange', s=20)
    ax.plot([mn,mx],[mn,mx],'r--',lw=2)
    ax.set_title(f'Hybrid  RMSE={results["Hybrid"]["RMSE"]:.2f}  α={alpha:.3f}')
    ax.set_xlabel('Actual RUL'); ax.set_ylabel('Predicted RUL'); ax.grid(alpha=0.3)
    ax = axes[1,2]
    res = y_test - pred_hybrid
    ax.hist(res, bins=40, color='darkorange', edgecolor='white', alpha=0.8)
    ax.axvline(0,          color='red',  lw=2, linestyle='--')
    ax.axvline(res.mean(), color='blue', lw=2, linestyle='--',
               label=f'Mean={res.mean():.2f}')
    ax.set_title('Hybrid Residuals')
    ax.set_xlabel('Residual'); ax.set_ylabel('Count')
    ax.legend(); ax.grid(alpha=0.3)
    plt.tight_layout()
    path = f'results_{fd}.png'
    plt.savefig(path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  ✅ Plot saved → {path}")

# ── 10. Main Loop ─────────────────────────────────────────────
all_results = {}
os.makedirs('saved_models', exist_ok=True)

for fd, cfg in DATASET_CFG.items():
    print(f"\n{'▓'*60}")
    print(f"  DATASET : {fd}  |  clusters={cfg['n_clusters']}  "
          f"epochs={cfg['epochs']}  xgb_trees={cfg['xgb_trees']}")
    print(f"{'▓'*60}")
    try:
        train_df, test_df, rul_df = load_data(fd)
    except FileNotFoundError as e:
        print(f"  ⚠️  Skipping {fd}: {e}")
        continue
    train_df, test_df = cluster_operating_conditions(
        train_df, test_df, cfg['n_clusters'])
    train_df = engineer_features(train_df)
    test_df  = engineer_features(test_df)
    train_df, test_df, scaler_map, feature_cols = normalize_by_cluster(
        train_df, test_df)
    X_seq, y_seq = make_sequences(train_df, feature_cols)
    X_test_seq   = make_test_sequences(test_df, feature_cols)
    y_test        = rul_df['RUL'].values.clip(max=RUL_CAP).astype(np.float32)
    print(f"  Train sequences : {X_seq.shape}")
    print(f"  Test  sequences : {X_test_seq.shape}")
    print(f"  Features        : {len(feature_cols)}")
    dataset      = TensorDataset(torch.tensor(X_seq), torch.tensor(y_seq))
    holdout_size = int(0.05 * len(dataset))
    val_size     = int(0.10 * len(dataset))
    train_size   = len(dataset) - val_size - holdout_size
    train_ds, val_ds, holdout_ds = random_split(
        dataset, [train_size, val_size, holdout_size],
        generator=torch.Generator().manual_seed(42))
    train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                              pin_memory=True, num_workers=2)
    val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE,
                              pin_memory=True, num_workers=2)
    def ds_to_tensors(ds):
        Xt = torch.stack([ds[i][0] for i in range(len(ds))])
        yt = np.array([ds[i][1].item() for i in range(len(ds))])
        return Xt, yt
    val_X,  val_y  = ds_to_tensors(val_ds)
    hold_X, hold_y = ds_to_tensors(holdout_ds)
    val_loader_inf  = DataLoader(TensorDataset(val_X),
                                 batch_size=BATCH_SIZE, pin_memory=True)
    hold_loader_inf = DataLoader(TensorDataset(hold_X),
                                 batch_size=BATCH_SIZE, pin_memory=True)
    test_loader_inf = DataLoader(TensorDataset(torch.tensor(X_test_seq)),
                                 batch_size=BATCH_SIZE, pin_memory=True)
    print(f"  Train:{train_size}  Val:{val_size}  Holdout:{holdout_size}")
    print(f"\n  ── Training LSTM ({fd}) ──")
    lstm_model = LSTMModel(input_size=X_seq.shape[2])
    lstm_model, history = train_lstm(lstm_model, train_loader, val_loader,
                                     cfg['epochs'])
    pred_lstm_hold = predict_lstm(lstm_model, hold_loader_inf)
    pred_lstm_test = predict_lstm(lstm_model, test_loader_inf)
    print(f"\n  ── Training XGBoost ({fd}) ──")
    X_tab_train = get_tabular_features(X_seq)
    X_tab_val   = get_tabular_features(val_X.numpy())
    X_tab_hold  = get_tabular_features(hold_X.numpy())
    X_tab_test  = get_tabular_features(X_test_seq)
    xgb_model = XGBRegressor(
        max_depth=3, n_estimators=cfg['xgb_trees'],
        learning_rate=0.03, subsample=0.6, colsample_bytree=0.6,
        reg_alpha=1.0, reg_lambda=5.0, min_child_weight=10, gamma=0.1,
        tree_method='hist', device='cuda', random_state=42, n_jobs=-1,
        early_stopping_rounds=50)
    xgb_model.fit(X_tab_train, y_seq,
                  eval_set=[(X_tab_val, val_y)], verbose=200)
    print(f"  ✅ Best XGB iteration: {xgb_model.best_iteration}")
    pred_xgb_hold = xgb_model.predict(X_tab_hold)
    pred_xgb_test = xgb_model.predict(X_tab_test)
    print(f"\n  ── Tuning Alpha ({fd}) ──")
    res = minimize_scalar(
        lambda a: np.sqrt(np.mean(
            (hold_y - (a*pred_xgb_hold + (1-a)*pred_lstm_hold))**2)),
        bounds=(0,1), method='bounded')
    alpha       = res.x
    pred_hybrid = alpha * pred_xgb_test + (1-alpha) * pred_lstm_test
    print(f"  ✅ α={alpha:.4f}  Holdout RMSE={res.fun:.4f}")
    print(f"\n  ── Results ({fd}) ──")
    results = {
        'LSTM'   : evaluate(y_test, pred_lstm_test,  "LSTM Only"),
        'XGBoost': evaluate(y_test, pred_xgb_test,   "XGBoost Only"),
        'Hybrid' : evaluate(y_test, pred_hybrid,     f"Hybrid (α={alpha:.3f})"),
    }
    all_results[fd] = results
    plot_results(fd, history, results, y_test,
                 pred_lstm_test, pred_xgb_test, pred_hybrid, alpha)
    torch.save(lstm_model.state_dict(), f'saved_models/lstm_{fd}.pt')
    joblib.dump(xgb_model,              f'saved_models/xgb_{fd}.pkl')
    joblib.dump(scaler_map,             f'saved_models/scaler_{fd}.pkl')
    np.save(f'saved_models/features_{fd}.npy', np.array(feature_cols))
    np.save(f'saved_models/alpha_{fd}.npy',    np.array([alpha]))
    print(f"  ✅ Models saved for {fd}")
    del lstm_model
    torch.cuda.empty_cache()

# ── 11. Final Summary ─────────────────────────────────────────
print(f"\n{'═'*62}")
print(f"  FINAL SUMMARY — ALL DATASETS")
print(f"{'═'*62}")
print(f"  {'Dataset':<10} {'Model':<12} {'RMSE':>8} {'MAE':>8} {'R²':>8}")
print(f"  {'─'*58}")
for fd, res in all_results.items():
    best_model = min(res, key=lambda k: res[k]['RMSE'])
    for model, metrics in res.items():
        marker = ' ✅' if model == best_model else ''
        print(f"  {fd:<10} {model:<12} "
              f"{metrics['RMSE']:>8.4f} {metrics['MAE']:>8.4f} "
              f"{metrics['R2']:>8.4f}{marker}")
    print(f"  {'─'*58}")
print(f"{'═'*62}")

# ── 12. Cross-dataset summary plot ───────────────────────────
if len(all_results) > 1:
    fds    = list(all_results.keys())
    models = ['LSTM','XGBoost','Hybrid']
    colors = ['royalblue','seagreen','darkorange']
    x      = np.arange(len(fds))
    width  = 0.25
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    fig.suptitle('All FD Datasets — Model Comparison', fontsize=15, fontweight='bold')
    ax = axes[0]
    for i, (model, color) in enumerate(zip(models, colors)):
        vals = [all_results[fd][model]['RMSE'] for fd in fds]
        bars = ax.bar(x + i*width, vals, width, label=model,
                      color=color, edgecolor='white')
        for bar, v in zip(bars, vals):
            ax.text(bar.get_x()+bar.get_width()/2,
                    bar.get_height()+0.2, f'{v:.1f}',
                    ha='center', fontsize=8, fontweight='bold')
    ax.set_xticks(x + width); ax.set_xticklabels(fds)
    ax.set_title('RMSE by Dataset'); ax.set_ylabel('RMSE')
    ax.legend(); ax.grid(axis='y', alpha=0.3)
    ax = axes[1]
    for i, (model, color) in enumerate(zip(models, colors)):
        vals = [all_results[fd][model]['R2'] for fd in fds]
        bars = ax.bar(x + i*width, vals, width, label=model,
                      color=color, edgecolor='white')
        for bar, v in zip(bars, vals):
            ax.text(bar.get_x()+bar.get_width()/2,
                    bar.get_height()+0.005, f'{v:.2f}',
                    ha='center', fontsize=8, fontweight='bold')
    ax.set_xticks(x + width); ax.set_xticklabels(fds)
    ax.set_title('R² by Dataset'); ax.set_ylabel('R²')
    ax.set_ylim(0, 1.1); ax.legend(); ax.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig('summary_all_datasets.png', dpi=150, bbox_inches='tight')
    plt.show()
    print("✅ Summary plot saved → summary_all_datasets.png")

# ── 13. Download everything (Colab only) ─────────────────────
if _is_colab():
    from google.colab import files
    download_files = ['summary_all_datasets.png']
    for fd in all_results.keys():
        download_files += [
            f'results_{fd}.png',
            f'saved_models/lstm_{fd}.pt',
            f'saved_models/xgb_{fd}.pkl',
        ]
    for f in download_files:
        if os.path.exists(f):
            files.download(f)

print("\n🎉 All datasets complete!")
```
